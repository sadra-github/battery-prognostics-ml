```python
from pathlib import Path
import pandas as pd

csv_path = Path('../data/processed/your_cycle_data.csv')  # edit for your dataset
if csv_path.exists():
    df = pd.read_csv(csv_path)
    display(df.head())
    print('Shape:', df.shape)
    display(df.dtypes.to_frame('dtype'))
    display(df.isna().sum().to_frame('missing_count'))
else:
    print(f'Update csv_path to an existing file: {csv_path.resolve()}')
```
